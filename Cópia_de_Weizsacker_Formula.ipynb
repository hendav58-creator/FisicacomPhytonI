```python
#massas

u = 931.5 # (MeV/c**2)
mp = 938.3 # (MeV/c**2)
mn = 939.6 # (MeV/c**2)

#parametros de Wapstra: # (MeV/c**2)

aV = 15.85
aS = 18.34
aC =  0.71
aA = 92.86
aP = 11.46
```

```python
Z = int(input(f"Digite o número atômico: "))
A = int(input(f"Digite o número de massa: "))
```

```python
#Fórmula de Weiszacker
def func(Z, A):
  return Z*mp + (A-Z)*mn - aV*A - aS*A**(2/3) + aC*(Z**2)*(A**(-1/3)) + (aA*(Z-(A/2))**2)/A - aP*(A**(-1/2))*((-1**Z) + (-1**(A-Z)))/2

M = func(Z, A)

print(f"A massa atômica em MeV/c**2 é: {M}")
```

```python
Mu = M/u
print(f"A massa atômica em u.m.a é {Mu}")
```
